```python
missing_indices = np.random.choice(s.index, size=10, replace=False)

s[missing_indices] = np.nan

print(s.head(15))
```

```python
valid_indices = s.dropna().index
outlier_indices = np.random.choice(valid_indices, size=5, replace=False)

s[outlier_indices] = s[outlier_indices] * 3

print("선택된 이상치 인덱스 및 변경된 값:")
print(s[outlier_indices])
```

결측값 보정 방식 4가지 비교

```python
missing_data = s[s.isna()]

print("--- [결측 발생 날짜 목록] ---")
print(missing_data.index)

print("\n--- [결측 개수] ---")
print(len(missing_data))
```

```python
s_dropna = s.dropna()

s_fillna_zero = s.fillna(0)

s_fillna_mean = s.fillna(s.mean())

s_fillna_ffill = s.ffill()

print("--- (1) 결측 제거 ---")
print(s_dropna.head())
print("\n--- (2) 0으로 채움 ---")
print(s_fillna_zero.head())
print("\n--- (3) 전체 평균으로 채움 ---")
print(s_fillna_mean.head())
print("\n--- (4) 바로 앞 날짜 값으로 채움 ---")
print(s_fillna_ffill.head())
```

```python
comparison_df = pd.DataFrame({
    'Mean': [
        s_dropna.mean(), 
        s_fillna_zero.mean(), 
        s_fillna_mean.mean(), 
        s_fillna_ffill.mean()
    ],
    'Std': [
        s_dropna.std(), 
        s_fillna_zero.std(), 
        s_fillna_mean.std(), 
        s_fillna_ffill.std()
    ]
}, index=['결측 제거 (dropna)', '0으로 채움 (fillna 0)', '전체 평균으로 채움 (fillna mean)', '앞 날짜 값으로 채움 (ffill)'])

print(comparison_df)
```

이상치 탐지와 보정

```python
if 's_fillna_ffill' in locals():
    s_clean = s_fillna_ffill.copy()
else:
   
    try:
        s_clean = s.ffill()
    except NameError:
        
        date_range = pd.date_range(start='2026-01-01', periods=100, freq='D')
        np.random.seed(42)
        s = pd.Series(np.random.normal(20, 5, 100), index=date_range)
        s_clean = s.ffill()

mean_val = s_clean.mean()
std_val = s_clean.std()

upper_bound = mean_val + (2 * std_val)
lower_bound = mean_val - (2 * std_val)

print(f"평균 (Mean): {mean_val:.4f}")
print(f"표준편차 (Std): {std_val:.4f}")
print(f"하한 (Lower Bound): {lower_bound:.4f}")
print(f"상한 (Upper Bound): {upper_bound:.4f}")
```

```python
outliers_detected = s_clean[(s_clean > upper_bound) | (s_clean < lower_bound)]

print("--- [임계치(평균 ± 2σ)를 벗어난 감지된 이상치] ---")
print(outliers_detected)
print(f"\n감지된 이상치 개수: {len(outliers_detected)}개")

if 'outlier_indices' in locals():
    print("\n--- [①에서 주입한 실제 이상치 인덱스] ---")
    print(outlier_indices)
    
    matched = set(outliers_detected.index) == set(outlier_indices)
    print(f"\n주입한 5일과 일치 여부: {matched}")
else:
    print("\n(참고: 'outlier_indices' 변수가 세션에 없어 직접 비교를 수행할 수 없습니다.)")
```

```python
cleaned_values = np.where(
    s_clean > upper_bound, 
    upper_bound, 
    np.where(s_clean < lower_bound, lower_bound, s_clean)
)

clean = pd.Series(cleaned_values, index=s_clean.index)

print("--- [이상치가 상·하한값으로 보정된 clean Series 확인] ---")
print(clean.head(15))
```

```python
print("--- [보정 전 (s_clean) describe] ---")
print(s_clean.describe())

print("\n--- [보정 후 (clean) describe] ---")
print(clean.describe())
```

 월별 집계와 이동평균

```python
monthly_summary = clean.groupby(clean.index.to_period('M')).agg(
    관측일수=('count'),
    평균=('mean'),
    최저=('min'),
    최고=('max')
)

print("--- [월별 관측일수·평균·최저·최고 요약] ---")
print(monthly_summary)
```

```python
rolling_mean_7 = clean.rolling(window=7).mean()

print("--- [7일 이동평균 결과 (앞부분 확인)] ---")
print(rolling_mean_7.head(10))
```

```python
top_5_days = clean.nlargest(5)

print("--- [기온 상위 5일 및 값] ---")
print(top_5_days)
```

```python
days_above_mean = clean[clean > mean_val]

print(f"전체 평균({mean_val:.4f})보다 기온이 높았던 날의 수: {len(days_above_mean)}일")
print("\n--- [평균 초과 날짜 및 기온 (상위 일부 확인)] ---")
print(days_above_mean.head(10))
```

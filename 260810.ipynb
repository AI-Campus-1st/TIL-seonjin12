```python
#Q6
SELECT COUNT(*) AS product_count, 
       AVG(unit_price) AS avg_unit_price 
FROM tb_product  
ORDER BY avg_unit_price DESC;
```

```python
#Q7
SELECT grade, 
       COUNT(*) AS customer_count, 
       ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM tb_customer), 2) AS percentage 
FROM tb_customer 
GROUP BY grade;
```

```python
Q8
SELECT DATE_FORMAT(order_dt, '%Y-%m') AS order_month, 
       COUNT(*) AS order_count 
FROM tb_order 
WHERE status != 'CANCEL' 
GROUP BY order_month 
ORDER BY order_month;
```

```python
Q9
SELECT category_id, 
       SUM(stock_qty) AS total_stock 
FROM tb_product 
GROUP BY category_id 
HAVING total_stock < 100;
```

```python
Q10
SELECT MAX(unit_price) - MIN(unit_price) AS price_diff 
FROM tb_product;
```

[조인] Q11~Q15

```python
Q11
SELECT c.customer_name, 
       o.order_dt, 
       SUM(oi.qty * oi.unit_price) AS total_amount
FROM tb_order o
JOIN tb_customer c ON o.customer_id = c.customer_id
JOIN tb_order_item oi ON o.order_id = oi.order_id
GROUP BY o.order_id, c.customer_name, o.order_dt;
```

```python
Q12
SELECT c.* 
FROM tb_customer c
LEFT JOIN tb_order o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
```

```python
Q13
SELECT p.* 
FROM tb_product p
LEFT JOIN tb_order_item oi ON p.product_id = oi.product_id
WHERE oi.product_id IS NULL;
```

```python
Q14
SELECT c.customer_name, 
       SUM(oi.qty * oi.unit_price) AS total_purchase
FROM tb_order o
JOIN tb_customer c ON o.customer_id = c.customer_id
JOIN tb_order_item oi ON o.order_id = oi.order_id
WHERE o.status != 'CANCEL'
GROUP BY c.customer_id, c.customer_name
ORDER BY total_purchase DESC
LIMIT 5;
```

```python
Q15
SELECT c.country, 
       p.category_id, 
       SUM(oi.qty * oi.unit_price) AS total_sales
FROM tb_order o
JOIN tb_customer c ON o.customer_id = c.customer_id
JOIN tb_order_item oi ON o.order_id = oi.order_id
JOIN tb_product p ON oi.product_id = p.product_id
WHERE o.status != 'CANCEL'
GROUP BY c.country, p.category_id
ORDER BY c.country, total_sales DESC;
```

```python
Q16
SELECT category_id, 
       COUNT(*) AS product_count,
       CASE 
           WHEN COUNT(*) > 3 THEN '많음' 
           ELSE '적음' 
       END AS product_status
FROM tb_product 
GROUP BY category_id;SELECT order_range, 
       COUNT(*) AS customer_count
FROM (
    SELECT c.customer_id,
           CASE 
               WHEN COUNT(o.order_id) = 0 THEN '0건'
               WHEN COUNT(o.order_id) BETWEEN 1 AND 2 THEN '1~2건'
               ELSE '3건 이상'
           END AS order_range
    FROM tb_customer c
    LEFT JOIN tb_order o ON c.customer_id = o.customer_id
    GROUP BY c.customer_id
) t
GROUP BY order_range
ORDER BY FIELD(order_range, '0건', '1~2건', '3건 이상');
```

```python
Q18
SELECT YEAR(o.order_dt) AS order_year,
       SUM(CASE WHEN MONTH(o.order_dt) = 1 THEN oi.qty * oi.unit_price ELSE 0 END) AS '1월',
       SUM(CASE WHEN MONTH(o.order_dt) = 2 THEN oi.qty * oi.unit_price ELSE 0 END) AS '2월',
       SUM(CASE WHEN MONTH(o.order_dt) = 3 THEN oi.qty * oi.unit_price ELSE 0 END) AS '3월',
       SUM(CASE WHEN MONTH(o.order_dt) = 4 THEN oi.qty * oi.unit_price ELSE 0 END) AS '4월',
       SUM(CASE WHEN MONTH(o.order_dt) = 5 THEN oi.qty * oi.unit_price ELSE 0 END) AS '5월',
       SUM(CASE WHEN MONTH(o.order_dt) = 6 THEN oi.qty * oi.unit_price ELSE 0 END) AS '6월',
       SUM(CASE WHEN MONTH(o.order_dt) = 7 THEN oi.qty * oi.unit_price ELSE 0 END) AS '7월',
       SUM(CASE WHEN MONTH(o.order_dt) = 8 THEN oi.qty * oi.unit_price ELSE 0 END) AS '8월',
       SUM(CASE WHEN MONTH(o.order_dt) = 9 THEN oi.qty * oi.unit_price ELSE 0 END) AS '9월',
       SUM(CASE WHEN MONTH(o.order_dt) = 10 THEN oi.qty * oi.unit_price ELSE 0 END) AS '10월',
       SUM(CASE WHEN MONTH(o.order_dt) = 11 THEN oi.qty * oi.unit_price ELSE 0 END) AS '11월',
       SUM(CASE WHEN MONTH(o.order_dt) = 12 THEN oi.qty * oi.unit_price ELSE 0 END) AS '12월'
FROM tb_order o
JOIN tb_order_item oi ON o.order_id = oi.order_id
WHERE o.status != 'CANCEL'
GROUP BY YEAR(o.order_dt)
ORDER BY order_year;
```

```python
Q19
SELECT 
    c.category_name, 
    p.product_name, 
    IFNULL(sold_stats.total_sold, 0) AS 판매수량, 
    p.stock_qty AS 남은재고,
    ROUND(IFNULL(sold_stats.total_sold, 0) / (IFNULL(sold_stats.total_sold, 0) + p.stock_qty) * 100, 1) AS 소진율
FROM tb_product p
LEFT JOIN (
    -- 취소를 제외한 상품별 판매 수량 집계
    SELECT 
        oi.product_id, 
        SUM(oi.qty) AS total_sold
    FROM tb_order_item oi
    JOIN tb_order o ON oi.order_id = o.order_id
    WHERE o.status != 'CANCEL'
    GROUP BY oi.product_id
) sold_stats ON p.product_id = sold_stats.product_id
JOIN tb_category c ON p.category_id = c.category_id
ORDER BY 소진율 DESC;
```

```python
Q20
SELECT 
    c.category_name, 
    p.product_name, 
    IFNULL(sold_stats.total_sold, 0) AS 판매수량, 
    p.stock_qty AS 남은재고,
    ROUND(IFNULL(sold_stats.total_sold, 0) / (IFNULL(sold_stats.total_sold, 0) + p.stock_qty) * 100, 1) AS 소진율
FROM tb_product p
LEFT JOIN (
    -- 취소를 제외한 상품별 판매 수량 집계
    SELECT 
        oi.product_id, 
        SUM(oi.qty) AS total_sold
    FROM tb_order_item oi
    JOIN tb_order o ON oi.order_id = o.order_id
    WHERE o.status != 'CANCEL'
    GROUP BY oi.product_id
) sold_stats ON p.product_id = sold_stats.product_id
JOIN tb_category c ON p.category_id = c.category_id
ORDER BY 소진율 DESC;
```

```python
Q21
SELECT 
    IFNULL(c.grade, '미지정') AS 등급,
    COUNT(DISTINCT c.customer_id) AS 인원수,
    COUNT(DISTINCT o.customer_id) AS 주문한인원수,
    IFNULL(SUM(CASE WHEN o.status != 'CANCEL' THEN oi.qty * oi.unit_price ELSE 0 END), 0) AS 총매출,
    ROUND(IFNULL(SUM(CASE WHEN o.status != 'CANCEL' THEN oi.qty * oi.unit_price ELSE 0 END), 0) / COUNT(DISTINCT c.customer_id), 2) AS '1인당 평균구매액'
FROM tb_customer c
LEFT JOIN tb_order o ON c.customer_id = o.customer_id
LEFT JOIN tb_order_item oi ON o.order_id = oi.order_id
GROUP BY IFNULL(c.grade, '미지정')
ORDER BY '1인당 평균구매액' DESC;
```

```python
CREATE DATABASE IF NOT EXISTS classmate_db DEFAULT CHARACTER SET utf8mb4 COLLATE utf8mb4_unicode_ci;
USE classmate_db;
```

```python
#회원
CREATE TABLE members (
    member_no INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) NOT NULL,
    phone VARCHAR(20),
    joined_at DATETIME NOT NULL,
    birth_yesr SMALLINT,
    gender CHAR(1)
) ENGINE=InnoDB;
```

```python
#카테고리
CREATE TABLE categories (
    category_code VARCHAR(20) PRIMARY KEY,
    category_name VARCHAR(50) NOT NULL,
    parent_code VARCHAR(20),
    sort_order SMALLINT NOT NULL,
    created_at DATETIME NOT NULL
) ENGINE=InnoDB;    
```

```python
#강사
CREATE TABLE instructors (
    instructor_id INT AUTO_INCREMENT PRIMARY KEY,
    bil TEXT,
    profile_img VARCHAR(255),
    created_at DATETIME NOT NULL,
    updated_at DATETIME NOT NULL
) ENGINE=InnoDB;
```

```python
#강의
CREATE TABLE courses (
    course_id VARCHAR(28) PRIMARY KEY,
    title VARCHAR(150) NOT NULL,
    category_code VARCHAR(20) NOT NULL,
    instructor INT NOT NULL,
    price INT UNSIGNED NOT NULL,
    sale_price INT UNSIGNED,
    course_level ENUM('BEGINNER', 'INTERMEDIATE', 'ADVANCED') NOT NULL,
    total_minutes INT UNSIGNED NOT NULL,
    published_at DATE,
    status ENUM('DRAFT', 'PUBLISHED', 'CLOSED') NOT NULL,
    FOREIGN KEY (category_code) REFERENCES categories(category_code),
    FOREIGN KEY (instructor) REFERENCES instructors(instructor_id)
) ENGINE=InnoDB;
```

```python
#강의 섹션
CREATE TABLE course_sections (
    section_id INT,
    course_id VARCHAR(20),
    section_no SMALLINT NOT NULL,
    title VARCHAR(100) NOT NULL,
    created_at DATETIME NOT NULL,
    PRIMARY KEY (section_id, course_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id) ON DELETE CASCADE
) ENGINE=InnoDB;
```

```python
#결제
CREATE TABLE payments (
    payment_id INT PRIMARY KEY,
    member_no INT NOT NULL,
    payment_method VARCHAR(20) NOT NULL,
    original_price INT NOT NULL,
    discount_amount INT NOT NULL,
    coupon_code VARCHAR(30),
    FOREIGN KEY (member_no) REFERENCES members(member_no)
) ENGINE=InnoDB;
```

```python
#수강신청
CREATE TABLE enrollments (
    enrollment_id INT PRIMARY KEY,
    member_no INT NOT NULL,
    course_id VARCHAR(20) NOT NULL,
    payment_id INT,
    enrolled_at DATETIME NOT NULL,
    progress_rate INT,
    FOREIGN KEY (member_no) REFERENCES members(member_no),
    FOREIGN KEY (course_id) REFERENCES courses(course_id),
    FOREIGN KEY (payment_id) REFERENCES payments(payment_id)
) ENGINE=InnoDB;
```

```python
#수강 후기
CREATE TABLE course_reviews (
    review_id VARCHAR(20) PRIMARY KEY,
    enrollment_id INT NOT NULL,
    member_no INT NOT NULL,
    course_id VARCHAR(20) NOT NULL,
    rating INT NOT NULL,
    FOREIGN KEY (enrollment_id) REFERENCES enrollments(enrollment_id),
    FOREIGN KEY (member_no) REFERENCES members(member_no),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
) ENGINE=InnoDB;
```

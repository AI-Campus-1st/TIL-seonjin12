```python
#문항1
word = 'apple'
vowels = 'aeiou'
count = 0
```

```python
for char in word:
    if vowels.find(char) != -1:
        count += 1

print(word, '# =>', count )
```

```python
word2 = 'banana'
count2 = 0
```

```python
for char in word2:
    if vowels.find(char) != -1:
        count2 += 1
print(word2, '# =>', count2)
```

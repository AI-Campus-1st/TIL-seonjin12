```python
# 문항2
scores = {
    'python': 80,
    'algorithm': 90,
    'django': 89,
    'web': 83,
} 
```

```python
print("subject(keys):",  scores.keys())
print("scores(values)", scores.values())
```

```python
총합 = sum(scores.values())
과목수 = len(scores)

average = 총합/과목수
print(average)
```

```python
#문항3

bloods = [
    'A', 'B', 'A', 'O', 'AB', 'AB',
    'O', 'A', 'B', 'O', 'B', 'AB',
]
```

```python
result = {}
#key는 혈액형, value는 개수
for blood in bloods:
    if blood in result:
        result[blood] += 1
    else:
        result[blood] = 1

print(result)
```

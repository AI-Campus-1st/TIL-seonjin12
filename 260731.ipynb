```python
#문항 1

import csv

input_file = 'scores.csv'
output_file = 'result.csv'

results = []

with open(input_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader) 
    

    subjects = ['kor', 'eng', 'math']
    
    for row in reader:
        name = row[0]
        scores = []
        
        for i in range(1, 4):
            val = row[i].strip()
            col_name = subjects[i - 1]
            
            try:
                if val == '':
                    raise ValueError("empty")
                score = float(val)
                scores.append(score)
            except ValueError:
              
                print(f"[처리] {name} - {col_name} 값 오류 ({val}) → 0점 처리")
                scores.append(0.0)
        
       
        avg = round(sum(scores) / len(scores), 1)
        results.append([name, avg])


with open(output_file, 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
   
    writer.writerow(['name', 'average'])
    
    writer.writerows(results)

print(f"완료: {output_file} 생성")
```

```python
#문항 2

import csv
import json

input_file = 'members.csv'
output_json = 'members.json'
output_log = 'errors.log'

success_count = 0
fail_count = 0

members_data = []
error_logs = []

with open(input_file, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader) 
    
  
    for idx, row in enumerate(reader, start=2):
        row_id, name, age = row[0], row[1].strip(), row[2]
        
       
        if not name:
            fail_count += 1
            error_logs.append(f"{idx}행 : name 비어있음")
            continue
            
      
        try:
            int_age = int(age)
        except ValueError:
            fail_count += 1
            error_logs.append(f"{idx}행 : invalid literal for int() with base 10: '{age}'")
            continue
            
       
        members_data.append({
            "id": int(row_id),
            "name": name,
            "age": int_age
        })
        success_count += 1


with open(output_json, 'w', encoding='utf-8') as f:
    json.dump(members_data, f, ensure_ascii=False, indent=4)


with open(output_log, 'w', encoding='utf-8') as f:
    for log in error_logs:
        f.write(log + '\n')


print(f"성공 {success_count}건 / 실패 {fail_count}건")
```

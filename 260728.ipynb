```python
#문항1

def get_middle_char(list_a):
    length = len(list_a)
    mid = length // 2

    if length % 2 == 1: 
        return list_a[mid]

    else:
        return list_a[mid - 1:mid + 1]
```

```python
print(get_middle_char('study'))
print(get_middle_char('python'))
```

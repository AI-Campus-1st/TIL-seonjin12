```python
#문항 2

# 반지름: 3, x좌표: 2, y좌표: 4
c1 = Circle(3, 2, 4)

c1.area() #=> 28.26
c1.circumference() #=> 18.84
c1.center() #=> (2, 4)
```

```python
class Circle:
    def __init__(self, r, x, y):
        self.r = r
        self.x = x
        self.y = y

    def area(self):
        return 3.14 * (self.r ** 2)

    def circumference(self):
        return 3.14 * 2 * self.r

    def center(self):
        return (self.x, self.y)

c1 = Circle(3, 2, 4)

print(c1.area())
print(c1.circumference())
print(c1.center())
    
```

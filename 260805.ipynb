```python
#문항 1

import numpy as np

scores_1d = [88, 92, 79, 95,
             75, 68, 82, 70,
             92, 99, 95, 88,
             60, 72, 65, 58,
             85, 80, 78, 90]

scores = np.array(scores_1d).reshape(5, 4)
scores = np.random.randint(50, 101, size= 20).reshape(5, 4)
```

```python
print("1.", scores, scores.shape, scores.ndim, scores.size, scores.dtype)
print("2. 학생평균", scores.mean(axis=1), "과목평균", scores.mean(axis=0))
print("3. 최고점", scores.max(), "위치:", np.unravel_index(scores.argmax(), scores.shape))
print("4. 전체평균", scores.mean(), "표준편차", scores.std(), "분산", scores.var())

m = scores.mean(axis=1) >= 80
print("5. 80점 이상의 학생 수", m.sum(), "성적", scores[m])
print("6. 90점 이상의 개수", (scores >= 90).sum(), "100점이 있는가.", (scores >= 100).any(), "60점 이상", (scores >= 60).all())
print("7. 등급", np.select([scores >= 90, scores >= 80], ['A', 'B'], default = ['C']  ))
      
```

```python
#문항 3
import numpy as np

sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가
```

```python
revenue = sales * price
print("1. 제품별 매출액:", revenue)
```

```python
total_revenue = sales @ price  
product_total_sales = sales.sum(axis=0)
print("2. 지점별 총매출:", total_revenue, "/ 제품별 총판매량:", product_total_sales)
```

```python
mean_sales = sales.mean(axis=0)
diff = sales - mean_sales
print("3. 평균 대비 편차 행렬:", diff)
```

```python
norm = (sales - sales.mean(axis=1, keepdims=True)) / sales.std(axis=1, keepdims=True)
print("4. 정규화 결과:", norm)
print("   행별 평균:", norm.mean(axis=1), "/ 행별 표준편차:", norm.std(axis=1))
```

```python
print("5. 전치행렬:\n", sales.T, "/ shape:", sales.T.shape)
```

```python
A = sales[:, :3]
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
identity_check = A @ inv_A
print("6. 행렬식:", det_A)
print("   역행렬:", inv_A)
print("   단위행렬 확인:", identity_check)
```

```python
max_idx = total_revenue.argmax()
cumulative_revenue = np.cumsum(total_revenue)
print("7. 최고 매출 지점 인덱스:", max_idx, "누적 매출:", cumulative_revenue)
```
